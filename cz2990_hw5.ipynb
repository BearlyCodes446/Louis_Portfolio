## Q2. getBondPrice
Excel targets: ppy=1 -> **2,170,604.06**, ppy=2 -> **2,171,686.39**

```python
# Use numpy
# No Loops!  All vectorized code.
def getBondPrice(y, face, couponRate, m, ppy=1):
    n = m * ppy
    t = np.arange(1,n+1)
    df = (1+y/ppy)**-t
    cf = np.full(n, couponRate*face/ppy)
    cf[-1] += face  #add up fv to cashflow at the end

    return(np.sum(cf * df))
```

```python
# Test values
y = 0.03
face = 2000000
couponRate = 0.04
m = 10
print(getBondPrice(y, face, couponRate, m, ppy=1))   # 2170604.06
print(getBondPrice(y, face, couponRate, m, ppy=2))   # 2171686.39
print(getBondPrice(y, face, couponRate, m))          # default ppy=1
```

## Q3. getBondDuration
Excel target: ppy=1 -> **8.5087** years

```python
# Use numpy
# No Loops!  All vectorized code.
def getBondDuration(y, face, couponRate, m, ppy=1):
    n = m * ppy
    t = np.arange(1,n+1)
    df = (1+y/ppy)**-t
    cf = np.full(n, couponRate*face/ppy)
    cf[-1] += face  #add up fv to cashflow at the end

    bondPrice = np.sum(cf * df)
    duration = np.sum(t/ppy * cf * df) / bondPrice
    return(duration)
```

```python
# Test values
y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1
print(getBondDuration(y, face, couponRate, m, ppy))   # 8.5087
```

## Q4. FizzBuzz 

```python
# Have this code run for any start and finish values
# Use numpy
# No Loops!  All vectorized code.
def FizzBuzz(start, finish):
    num = np.arange(start, finish+1)
    obj = np.array(num,dtype=object)
    mask = num % 3 == 0
    obj[mask] = "fizz"
    mask = num % 5 == 0
    obj[mask] = "buzz"
    mask = (num % 3 == 0) & (num % 5 == 0)
    obj[mask] = "fizzbuzz"
    return(obj)
```

```python
print(FizzBuzz(1, 15))
print(FizzBuzz(67, 96))  
```
