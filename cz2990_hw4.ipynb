```python
def getBondPrice(y,face,couponRate,m,ppy=1):
    bondPrice = 0
    n = m * ppy                      #number of payment throughout the bond's life
    coupon= face * couponRate / ppy  #coupon paid each period
    for t in range(1, n+1):          #the for loop 
        df = 1/(1 + y/ppy)**t        #discount factor
        cf = coupon                  #cash flow for this period
        if t == n:                   #if at last period, add the FV to CF
            cf += face
        bondPrice += cf * df
    return(bondPrice)
```

```python
# Test values
y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1
ppy = 2
#<no ppy value passed>
print(getBondPrice(y, face, couponRate, m, ppy=1))  #pay once per year
print(getBondPrice(y, face, couponRate, m, ppy=2))  #higher frequency of payments, lower bond price
print(getBondPrice(y, face, couponRate, m))
```

```python
def getBondDuration(y, face, couponRate, m, ppy = 1):
    bondPrice = 0
    weightedSum = 0
    n = m * ppy                      
    coupon= face * couponRate / ppy  #coupon paid each period
    for t in range(1, n+1):          #the for loop 
        df = 1/(1 + y/ppy)**t        #discount factor
        cf = coupon                  #cash flow for this period
        if t == n:                   #if at last period, add the FV to CF
            cf += face
        bondPrice += cf * df
        weightedSum += (t/ppy) * cf * df   #time in years * pv
    bondDuration = weightedSum / bondPrice

    return(bondDuration)
```

```python
# Test values

y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1

print(getBondDuration(y, face, couponRate, m, ppy=1))  #pay once per year
```

```python
def getBondPrice_E(face, couponRate, m, yc):
    #<Your work using enumerate>
    bondPrice = 0
    coupon = face * couponRate
    for i, y in enumerate(yc):
        t = i + 1                   # i = position, y = that year's rate,  convert posiiton to year by adding 1
        df = 1/(1 + y)**t          
        cf = coupon                 
        if t == m:
            cf += face
        bondPrice += cf * df
    return(bondPrice)
```

```python
def getBondPrice_Z(face, couponRate, times, yc):
    maturity = times[-1]     # -1 = last element → 7
    bondPrice = 0
    coupon = face * couponRate
    for t, y in zip(times, yc):
        df = 1/(1 + y)**t
        cf = coupon
        if t == maturity: 
            cf += face
        bondPrice += cf * df
    return(bondPrice)
```

```python
yc = [.010,.015,.020,.025,.030]
times = [1,1.5,3,4,7]
face = 2000000
couponRate = .04
print(getBondPrice_Z(face, couponRate, times, yc))   
```

```python
# Have this code run for any start and finish values

def FizzBuzz(start, finish):
    outlist = []
    for i in range(start, finish + 1):
        if i % 15 == 0:
            outlist.append('fizzbuzz')
        elif i % 3 == 0:
            outlist.append('fizz')
        elif i % 5 == 0:
            outlist.append('buzz')
        else:
            outlist.append(i)        # the number itself (an int, not a string)
    return(outlist)
```

```python
print(FizzBuzz(9, 2333))

print(FizzBuzz(10, 20))    # "any start and finish": not starting at 1
```
